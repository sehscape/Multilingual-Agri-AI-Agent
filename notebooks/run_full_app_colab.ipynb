1. `Runtime → Change runtime type → T4 GPU` (optional, makes it faster).
2. `Runtime → Run all`.
3. Wait ~5–10 min for the first run (installs + model downloads).
4. Click the `https://xxxx.gradio.live` link printed by the last cell.
5. Pick your language at the top of the page — the whole screen switches and the
   microphone listens in that language.

The link stays live **only while this notebook tab is open**. Good for a demo or
viva; for a permanent link use the free Render “Lite” deploy (see `DEPLOY.md`).

```python
# 4) Configure the FULL app
import os

os.environ['WHISPER_MODEL_ID']     = 'openai/whisper-small'        # speech-to-text
os.environ['AGENT_BACKEND']        = 'langchain'                   # LangChain ReAct agent with tool-calling
os.environ['USE_STUB_LLM']         = 'false'                       # a real open-source LLM drives the agent
os.environ['USE_HF_INFERENCE_API'] = 'false'                       # run it locally here
os.environ['LOCAL_LLM_MODEL_ID']   = 'Qwen/Qwen2.5-1.5B-Instruct'  # bigger = better (Colab can handle it)
os.environ['USE_STUB_RAG']         = 'false'                       # real semantic scheme search
os.environ['TTS_ENGINE']           = 'gtts'                        # real Hi/Mr/Pa voice
os.environ['DEV_MODE']             = 'true'                        # show the pipeline trace

# --- OPTIONAL: answers in Hindi / Punjabi / Marathi (needs a Hugging Face token
#     with IndicTrans2 access and a ~4 GB download) ---
# os.environ['USE_STUB_TRANSLATION'] = 'false'
os.environ.setdefault('USE_STUB_TRANSLATION', 'true')  # default: answers in English
```
